```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
df = sns.load_dataset('titanic')
```

```python
display(df.head())
sns.set(style="whitegrid")
```

```python
# 1. Scatter Plot - Relationship between age and fare
plt.figure(figsize=(5, 5))
sns.scatterplot(data=df, x='age', y='fare', hue='survived', alpha=0.8)
plt.title('Scatter Plot: Age vs Fare')
plt.show()
```

```python
# 2. Line Plot - Trends in fare over age
plt.figure(figsize=(5, 5))
sns.lineplot(data=df, x='age', y='fare', hue='class')
plt.title('Line Plot: Age vs Fare by Class')
plt.show()
```

```python
# 3. Bar Plot - Average fare per class
plt.figure(figsize=(5, 5))
sns.barplot(data=df, x='class', y='fare')
plt.title('Bar Plot: Total Fare by Class')
plt.show()
```

```python
# 4. Histogram - Distribution of age
plt.figure(figsize=(8, 5))
sns.histplot(df['age'], bins=20, kde=True)
plt.title('Histogram: Age Distribution')
plt.show()
```

```python
# 5. Box Plot - Fare distribution across different classes
plt.figure(figsize=(5, 5))
sns.boxplot(data=df, x='class', y='fare', hue='survived')
plt.title('Box Plot: Fare by Class and Survival')
plt.show()
```

```python
# 6. Violin Plot - Age distribution by class
plt.figure(figsize=(5, 5))
sns.violinplot(data=df, x='class', y='age', hue='sex', split=True)
plt.title('Violin Plot: Age Distribution by Class and Sex')
plt.show()
```

```python
# 7. Pair Plot - Relationship between numerical variables
sns.pairplot(df[['age', 'fare', 'pclass', 'survived']].dropna(), hue='survived')
plt.suptitle('Pair Plot: Titanic Dataset', y=1.02)
plt.show()
```

```python
# 8. Heatmap - Correlation between numerical variables
plt.figure(figsize=(8, 5))
sns.heatmap(df[['age', 'fare', 'pclass', 'survived']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap: Feature Correlation')
plt.show()
```
