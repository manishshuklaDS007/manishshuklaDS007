```python
#Task 1: Create scatter plots for the following data
#a. Create scatter plots with between (mpg-hp, mpg-weight),
#and (hp-mpg, hp-time-to-60) with respect to each brand in the same plane
plt.figure(figsize=(12, 10))
plt.subplot (2,2,1)
sns.scatterplot (data=df, x="mpg", y="hp", hue="brand")
plt.title("mpg vs hp")

plt.subplot (2,2,2)
sns.scatterplot (data=df, x="mpg", y="weightlbs", hue="brand")
plt.title("mpg vs weight")

plt.subplot (2,2,3)
sns.scatterplot (data=df, x="hp", y="mpg", hue="brand")
plt.title("hp vs mpg")

plt.subplot (2,2,4)
sns.scatterplot (data=df, x="hp", y="time-to-60", hue="brand")
plt.title("hp vs time-to-60")
plt.title ("layout")
plt.show()
```
